## Q1. What is NLP and what libraries do we need?

**Answer:**  
NLP (**Natural Language Processing**) helps computers read and understand human language.  

For this chatbot we only need **3 libraries**:

| Library | Purpose |
|---|---|
| `pandas` | Store questions & answers in a table |
| `scikit-learn` | Convert text → numbers (TF-IDF) and measure similarity |

````python
# Step 1 — Import all required libraries
import pandas as pd
import pickle
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("✅ Libraries imported successfully")
````

## Q2. What is the chatbot dataset?

A chatbot dataset is a simple table of **questions** paired with **answers**.  
When a user types something, the bot finds the closest matching question and returns its answer.

Our dataset covers: Greetings · About the bot · NLP & ML concepts · Farewell

````python
# Step 2 — Create the Q&A dataset
questions = [
    # Greetings
    "hi", "hello", "hey there", "good morning", "good evening", "good night",
    # About the bot
    "what is your name", "who are you", "who made you",
    "how are you", "how are you doing",
    "what can you do", "how can you help me",
    # NLP concepts
    "what is nlp", "what is natural language processing",
    "what is tfidf", "what is cosine similarity",
    "what is a chatbot", "what is retrieval based chatbot",
    # ML / AI concepts
    "what is machine learning", "what is artificial intelligence",
    "what is deep learning", "what is data science",
    # Gratitude
    "thank you", "thanks a lot", "that was helpful",
    # Farewell
    "bye", "goodbye", "see you later",
]

answers = [
    # Greetings
    "Hello! How can I help you today? 😊",
    "Hi there! How can I help you today? 😊",
    "Hey! Good to see you.",
    "Good morning! Hope you have a great day. 🌞",
    "Good evening! How can I assist you?",
    "Good night! Sweet dreams. 🌙",
    # About the bot
    "I am Manna 🤖, a simple NLP chatbot built by Madhusudan Manna.",
    "I am Manna, your NLP chatbot assistant!",
    "I was built by Madhusudan Manna as a learning project on NLP.",
    "I am doing great, thanks for asking! 😊",
    "I am doing great, thanks for asking!",
    "I can answer questions using NLP techniques like TF-IDF and Cosine Similarity.",
    "Just ask me a question and I will try my best to answer it!",
    # NLP concepts
    "NLP (Natural Language Processing) is the field that helps computers read and understand human language.",
    "Natural Language Processing (NLP) enables machines to interpret, understand, and respond to human text.",
    "TF-IDF stands for Term Frequency–Inverse Document Frequency. It converts text into numbers so machines can compare sentences.",
    "Cosine Similarity measures how similar two text vectors are. A score of 1 means identical, 0 means completely different.",
    "A chatbot is a software program that can hold a text conversation with a human using rules or AI/NLP techniques.",
    "A retrieval-based chatbot finds the best matching pre-written answer from a dataset using similarity techniques.",
    # ML / AI concepts
    "Machine Learning is the ability of a computer to learn patterns from data without being explicitly programmed.",
    "Artificial Intelligence (AI) is the simulation of human intelligence in machines.",
    "Deep Learning is a subset of Machine Learning that uses neural networks with many layers to learn from large data.",
    "Data Science is the process of extracting insights and knowledge from structured and unstructured data.",
    # Gratitude
    "You are welcome! 😊",
    "You are most welcome!",
    "Happy to help! Let me know if you need anything else.",
    # Farewell
    "Goodbye! Have a nice day. 👋",
    "Bye! Take care. 👋",
    "See you soon! 👋",
]

df = pd.DataFrame({"question": questions, "answer": answers})
print(f"Dataset size: {len(df)} rows")
df.head()
````

---
## Q3. Why do we clean/preprocess text?

**Answer:**  
Computers treat `'Hi'`, `'HI!'`, and `'hi'` as three different words.  
Cleaning (lowercase + remove punctuation) makes them the **same word** — improving accuracy.

````python
# Step 3 — Text cleaning function
def clean_text(text):
    text = text.lower()                        # convert to lowercase
    text = re.sub(r"[^a-z0-9\s]", "", text)  # remove punctuation
    return text.strip()

# Apply to the question column
df["clean_question"] = df["question"].apply(clean_text)

# Show quick example
print("Original : 'What is NLP?'")
print("Cleaned  :", clean_text("What is NLP?"))
print()
df[["question", "clean_question"]].head()
````

---
## Q4. What is TF-IDF and how does it work?

**Answer:**  
**TF-IDF** converts each sentence into a row of numbers (a vector).
- **TF (Term Frequency):** How often a word appears in a sentence.
- **IDF (Inverse Document Frequency):** Reduces weight of common words like 'is', 'the' and highlights rare, important words.

Result: every question becomes a unique numeric fingerprint.

````python
# Step 4 — Convert text to TF-IDF vectors
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["clean_question"])  # fit = learn vocab, transform = convert

print(f"Number of questions : {X.shape[0]}")
print(f"Vocabulary size     : {X.shape[1]} unique words")
print(f"TF-IDF matrix shape : {X.shape}")
````

---
## Q5. How does the chatbot find the best answer?

**Answer:**  
When user types a message:
1. Clean & convert it to a TF-IDF vector
2. Compute **cosine similarity** against every question in the dataset
3. Return the answer of the question with the **highest similarity score**
4. If the score is too low (< 0.25), say we did not understand

````python
# Step 5 — Chatbot response function
def chatbot_response(user_input):
    cleaned    = clean_text(user_input)             # clean input
    user_vec   = vectorizer.transform([cleaned])    # convert to TF-IDF
    scores     = cosine_similarity(user_vec, X)     # compare with all questions
    best_idx   = scores.argmax()                    # index of highest score
    best_score = scores[0][best_idx]

    if best_score < 0.25:
        return "Sorry, I did not understand that. Can you rephrase? 🤔"
    return df["answer"][best_idx]

# Quick tests
for q in ["hello", "what is nlp", "what can you do", "bye", "what is deep learning"]:
    print(f"User  : {q}")
    print(f"Manna : {chatbot_response(q)}")
    print()
````

---
## Q6. How do we save the chatbot so we don't retrain every time?

**Answer:**  
We use Python's **pickle** module to save the vectorizer, TF-IDF matrix, and dataframe into one file.  
The Streamlit app loads this file and is ready to chat instantly without retraining.

````python
# Step 6 — Save the chatbot model (store only numpy arrays, not pandas objects)
import numpy as np

chatbot_model = {
    "vectorizer"  : vectorizer,
    "X"           : X,
    "questions"   : df["clean_question"].tolist(),   # plain Python list — no pandas dtype issues
    "answers"     : df["answer"].tolist()            # plain Python list — no pandas dtype issues
}

with open("Manna.pkl", "wb") as f:
    pickle.dump(chatbot_model, f)

print("✅ Chatbot saved as Manna.pkl")
````

---
## Q7. How do we verify the saved pickle file works?

**Answer:**  
We reload `Manna.pkl` from scratch and run a few sample questions to confirm it works.

````python
# Step 7 — Load and verify
with open("Manna.pkl", "rb") as f:
    loaded = pickle.load(f)

v         = loaded["vectorizer"]
lX        = loaded["X"]
l_answers = loaded["answers"]

def test_loaded(user_input):
    cleaned = clean_text(user_input)
    vec     = v.transform([cleaned])
    sim     = cosine_similarity(vec, lX)
    idx     = int(sim.argmax())
    if sim[0][idx] < 0.25:
        return "Sorry, I did not understand that."
    return l_answers[idx]

print("=== Loaded model test ===")
for q in ["good morning", "thanks", "what is cosine similarity", "who made you"]:
    print(f"User  : {q}")
    print(f"Manna : {test_loaded(q)}")
    print()

print("✅ Model loaded and verified successfully!")
````

"""
app.py  –  Manna ChatBot  (Streamlit version)
Loads Manna.pkl and answers user questions using TF-IDF + Cosine Similarity.
Run with:  streamlit run app.py
"""

import re
import pickle
import streamlit as st
from sklearn.metrics.pairwise import cosine_similarity

# ── Page config ────────────────────────────────────────────────────────────────
st.set_page_config(page_title="Manna ChatBot", page_icon="🤖", layout="centered")

# ── Load model (cached so it is read only once per session) ────────────────────
@st.cache_resource
def load_model():
    with open("Manna.pkl", "rb") as f:
        model = pickle.load(f)
    return model["vectorizer"], model["X"], model["df"]

vectorizer, X, df = load_model()

# ── Text cleaning (same as training) ──────────────────────────────────────────
def clean_text(text: str) -> str:
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", "", text)
    return text.strip()

# ── Response logic ─────────────────────────────────────────────────────────────
def chatbot_response(user_input: str) -> str:
    cleaned = clean_text(user_input)
    if not cleaned:
        return "Please type something so I can help you!"
    user_vec = vectorizer.transform([cleaned])
    similarity = cosine_similarity(user_vec, X)
    best_idx = similarity.argmax()
    best_score = similarity[0][best_idx]
    if best_score < 0.3:
        return "Sorry, I did not understand that. Can you rephrase?"
    return df["answer"].iloc[best_idx]

# ── UI ─────────────────────────────────────────────────────────────────────────
st.title("🤖 Manna ChatBot")
st.caption("A simple retrieval chatbot built with TF-IDF + Cosine Similarity")

# Keep conversation history in session state
if "messages" not in st.session_state:
    st.session_state.messages = []

# Display existing conversation
for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.markdown(msg["content"])

# Chat input at the bottom
user_input = st.chat_input("Type your message here…")

if user_input:
    # Show user message
    st.session_state.messages.append({"role": "user", "content": user_input})
    with st.chat_message("user"):
        st.markdown(user_input)

    # Generate and show bot reply
    reply = chatbot_response(user_input)
    st.session_state.messages.append({"role": "assistant", "content": reply})
    with st.chat_message("assistant"):
        st.markdown(reply)
